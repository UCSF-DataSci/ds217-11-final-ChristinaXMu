```python
## Setup

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb
import os

# Load prepared data from Q6
X_train = pd.read_csv('output/q6_X_train.csv')
X_test = pd.read_csv('output/q6_X_test.csv')
y_train = pd.read_csv('output/q6_y_train.csv').squeeze()  # Convert to Series
y_test = pd.read_csv('output/q6_y_test.csv').squeeze()

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")
```

```python
# ========================================
# STEP 2: TRAIN MODEL 1 - LINEAR REGRESSION
# ========================================
print("\n" + "="*60)
print("TRAINING MODEL 1: LINEAR REGRESSION")
print("="*60)

print("\nInitializing Linear Regression model...")
lr_model = LinearRegression()

print("Training Linear Regression...")
lr_model.fit(X_train, y_train)
print("✓ Model trained")

# Make predictions
print("Making predictions...")
lr_train_pred = lr_model.predict(X_train)
lr_test_pred = lr_model.predict(X_test)
print("✓ Predictions generated")

# Calculate metrics
print("\nCalculating metrics...")
lr_train_r2 = r2_score(y_train, lr_train_pred)
lr_test_r2 = r2_score(y_test, lr_test_pred)
lr_train_rmse = np.sqrt(mean_squared_error(y_train, lr_train_pred))
lr_test_rmse = np.sqrt(mean_squared_error(y_test, lr_test_pred))
lr_train_mae = mean_absolute_error(y_train, lr_train_pred)
lr_test_mae = mean_absolute_error(y_test, lr_test_pred)

print("\nLinear Regression Performance:")
print(f"  Train R²:   {lr_train_r2:.4f}")
print(f"  Test R²:    {lr_test_r2:.4f}")
print(f"  Train RMSE: {lr_train_rmse:.4f}")
print(f"  Test RMSE:  {lr_test_rmse:.4f}")
print(f"  Train MAE:  {lr_train_mae:.4f}")
print(f"  Test MAE:   {lr_test_mae:.4f}")
```

```python
# ========================================
# STEP 3: TRAIN MODEL 2 - XGBOOST
# ========================================
print("\n" + "="*60)
print("TRAINING MODEL 2: XGBOOST")
print("="*60)

print("\nInitializing XGBoost model...")
xgb_model = xgb.XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    verbosity=0  # Suppress warnings
)

print("Training XGBoost...")
xgb_model.fit(X_train, y_train)
print("✓ Model trained")

# Make predictions
print("Making predictions...")
xgb_train_pred = xgb_model.predict(X_train)
xgb_test_pred = xgb_model.predict(X_test)
print("✓ Predictions generated")

# Calculate metrics
print("\nCalculating metrics...")
xgb_train_r2 = r2_score(y_train, xgb_train_pred)
xgb_test_r2 = r2_score(y_test, xgb_test_pred)
xgb_train_rmse = np.sqrt(mean_squared_error(y_train, xgb_train_pred))
xgb_test_rmse = np.sqrt(mean_squared_error(y_test, xgb_test_pred))
xgb_train_mae = mean_absolute_error(y_train, xgb_train_pred)
xgb_test_mae = mean_absolute_error(y_test, xgb_test_pred)

print("\nXGBoost Performance:")
print(f"  Train R²:   {xgb_train_r2:.4f}")
print(f"  Test R²:    {xgb_test_r2:.4f}")
print(f"  Train RMSE: {xgb_train_rmse:.4f}")
print(f"  Test RMSE:  {xgb_test_rmse:.4f}")
print(f"  Train MAE:  {xgb_train_mae:.4f}")
print(f"  Test MAE:   {xgb_test_mae:.4f}")
```

```python
# ========================================
# STEP 4: TRAIN MODEL 3 - RANDOM FOREST (OPTIONAL)
# ========================================
print("\n" + "="*60)
print("TRAINING MODEL 3: RANDOM FOREST")
print("="*60)

print("\nInitializing Random Forest model...")
rf_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1,  # Use all CPU cores
    verbose=0
)

print("Training Random Forest...")
rf_model.fit(X_train, y_train)
print("✓ Model trained")

# Make predictions
print("Making predictions...")
rf_train_pred = rf_model.predict(X_train)
rf_test_pred = rf_model.predict(X_test)
print("✓ Predictions generated")

# Calculate metrics
print("\nCalculating metrics...")
rf_train_r2 = r2_score(y_train, rf_train_pred)
rf_test_r2 = r2_score(y_test, rf_test_pred)
rf_train_rmse = np.sqrt(mean_squared_error(y_train, rf_train_pred))
rf_test_rmse = np.sqrt(mean_squared_error(y_test, rf_test_pred))
rf_train_mae = mean_absolute_error(y_train, rf_train_pred)
rf_test_mae = mean_absolute_error(y_test, rf_test_pred)

print("\nRandom Forest Performance:")
print(f"  Train R²:   {rf_train_r2:.4f}")
print(f"  Test R²:    {rf_test_r2:.4f}")
print(f"  Train RMSE: {rf_train_rmse:.4f}")
print(f"  Test RMSE:  {rf_test_rmse:.4f}")
print(f"  Train MAE:  {rf_train_mae:.4f}")
print(f"  Test MAE:   {rf_test_mae:.4f}")
```

```python
# ========================================
# STEP 5: CHECK FOR DATA LEAKAGE
# ========================================
print("\n" + "="*60)
print("DATA LEAKAGE CHECK")
print("="*60)

print("\nChecking for warning signs of data leakage...")

# Warning flags
warnings = []

# Check for perfect or near-perfect scores
if xgb_test_r2 > 0.999:
    warnings.append(f"⚠️  XGBoost R² = {xgb_test_r2:.6f} (suspiciously perfect!)")
if rf_test_r2 > 0.999:
    warnings.append(f"⚠️  Random Forest R² = {rf_test_r2:.6f} (suspiciously perfect!)")

# Check for unrealistically low errors
if xgb_test_rmse < 0.01:
    warnings.append(f"⚠️  XGBoost RMSE = {xgb_test_rmse:.6f} (unrealistically low!)")
if rf_test_rmse < 0.01:
    warnings.append(f"⚠️  Random Forest RMSE = {rf_test_rmse:.6f} (unrealistically low!)")

# Check if train and test are too similar
xgb_diff = abs(xgb_train_r2 - xgb_test_r2)
rf_diff = abs(rf_train_r2 - rf_test_r2)
if xgb_diff < 0.01:
    warnings.append(f"⚠️  XGBoost train/test R² difference = {xgb_diff:.6f} (too similar!)")
if rf_diff < 0.01:
    warnings.append(f"⚠️  Random Forest train/test R² difference = {rf_diff:.6f} (too similar!)")

if warnings:
    print("\n⚠️  WARNING: POTENTIAL DATA LEAKAGE DETECTED!")
    print("\nWarning signs:")
    for warning in warnings:
        print(f"  {warning}")
    print("\nCommon causes:")
    print("  - Using rolling windows of the target variable")
    print("  - Features with correlation > 0.99 with target")
    print("  - Including target variable in features")
    print("\nAction needed:")
    print("  1. Review Q6 feature selection")
    print("  2. Check for circular prediction logic")
    print("  3. Remove problematic features and retrain")
else:
    print("✓ No obvious data leakage detected")
    print("  All metrics appear reasonable")
```

```python
# ========================================
# STEP 6: MODEL COMPARISON
# ========================================
print("\n" + "="*60)
print("MODEL COMPARISON")
print("="*60)

# Create comparison summary
print("\nTest Set Performance Comparison:")
print(f"{'Model':<20} {'R²':<10} {'RMSE':<10} {'MAE':<10}")
print("-" * 50)
print(f"{'Linear Regression':<20} {lr_test_r2:<10.4f} {lr_test_rmse:<10.4f} {lr_test_mae:<10.4f}")
print(f"{'XGBoost':<20} {xgb_test_r2:<10.4f} {xgb_test_rmse:<10.4f} {xgb_test_mae:<10.4f}")
print(f"{'Random Forest':<20} {rf_test_r2:<10.4f} {rf_test_rmse:<10.4f} {rf_test_mae:<10.4f}")

# Identify best model
best_model_name = max(
    [('Linear Regression', lr_test_r2),
     ('XGBoost', xgb_test_r2),
     ('Random Forest', rf_test_r2)],
    key=lambda x: x[1]
)[0]

print(f"\n✓ Best model (by R²): {best_model_name}")
```

```python
# ========================================
# STEP 7: EXTRACT FEATURE IMPORTANCE
# ========================================
print("\n" + "="*60)
print("FEATURE IMPORTANCE EXTRACTION")
print("="*60)

print("\nExtracting feature importance from XGBoost...")
feature_importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': xgb_model.feature_importances_
}).sort_values('importance', ascending=False)

print(f"\nTop 10 most important features:")
for i, row in feature_importance.head(10).iterrows():
    print(f"  {row['feature']:<30} {row['importance']:.4f}")

# Calculate cumulative importance
feature_importance['cumulative_importance'] = feature_importance['importance'].cumsum()
top3_importance = feature_importance.head(3)['importance'].sum()
print(f"\nTop 3 features account for {top3_importance*100:.2f}% of total importance")
```

```python
# ========================================
# SAVE Q7 ARTIFACT 1: q7_predictions.csv
# ========================================
print("\n" + "="*60)
print("SAVING Q7 ARTIFACTS")
print("="*60)

print("\nSaving predictions...")
predictions_df = pd.DataFrame({
    'actual': y_test.values,
    'predicted_linear': lr_test_pred,
    'predicted_xgboost': xgb_test_pred,
    'predicted_random_forest': rf_test_pred
})
predictions_df.to_csv('output/q7_predictions.csv', index=False)
print(f"✓ Saved: output/q7_predictions.csv ({len(predictions_df)} rows)")
```

```python
# ========================================
# SAVE Q7 ARTIFACT 2: q7_model_metrics.txt
# ========================================
print("\nSaving model metrics...")
with open('output/q7_model_metrics.txt', 'w') as f:
    f.write("MODEL PERFORMANCE METRICS\n")
    f.write("=" * 60 + "\n\n")
    
    f.write("LINEAR REGRESSION:\n")
    f.write(f"  Train R²:   {lr_train_r2:.4f}\n")
    f.write(f"  Test R²:    {lr_test_r2:.4f}\n")
    f.write(f"  Train RMSE: {lr_train_rmse:.4f}\n")
    f.write(f"  Test RMSE:  {lr_test_rmse:.4f}\n")
    f.write(f"  Train MAE:  {lr_train_mae:.4f}\n")
    f.write(f"  Test MAE:   {lr_test_mae:.4f}\n\n")
    
    f.write("XGBOOST:\n")
    f.write(f"  Train R²:   {xgb_train_r2:.4f}\n")
    f.write(f"  Test R²:    {xgb_test_r2:.4f}\n")
    f.write(f"  Train RMSE: {xgb_train_rmse:.4f}\n")
    f.write(f"  Test RMSE:  {xgb_test_rmse:.4f}\n")
    f.write(f"  Train MAE:  {xgb_train_mae:.4f}\n")
    f.write(f"  Test MAE:   {xgb_test_mae:.4f}\n\n")
    
    f.write("RANDOM FOREST:\n")
    f.write(f"  Train R²:   {rf_train_r2:.4f}\n")
    f.write(f"  Test R²:    {rf_test_r2:.4f}\n")
    f.write(f"  Train RMSE: {rf_train_rmse:.4f}\n")
    f.write(f"  Test RMSE:  {rf_test_rmse:.4f}\n")
    f.write(f"  Train MAE:  {rf_train_mae:.4f}\n")
    f.write(f"  Test MAE:   {rf_test_mae:.4f}\n\n")
    
    f.write(f"BEST MODEL (by Test R²): {best_model_name}\n")

print("✓ Saved: output/q7_model_metrics.txt")
```

```python
# ========================================
# SAVE Q7 ARTIFACT 3: q7_feature_importance.csv
# ========================================
print("\nSaving feature importance...")
feature_importance[['feature', 'importance']].to_csv('output/q7_feature_importance.csv', index=False)
print(f"✓ Saved: output/q7_feature_importance.csv ({len(feature_importance)} features)")
```

```python
# ========================================
# Q8: RESULTS
# ========================================
print("\n" + "="*60)
print("Q8: RESULTS")
print("="*60)

# ========================================
# STEP 1: CREATE FINAL VISUALIZATIONS
# ========================================
print("\nCreating final visualizations...")

# Create figure with 4 subplots
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(2, 2, hspace=0.3, wspace=0.3)

# Plot 1: Model Performance Comparison (top-left)
ax1 = fig.add_subplot(gs[0, 0])
models = ['Linear\nRegression', 'XGBoost', 'Random\nForest']
test_r2_scores = [lr_test_r2, xgb_test_r2, rf_test_r2]
test_rmse_scores = [lr_test_rmse, xgb_test_rmse, rf_test_rmse]

x = np.arange(len(models))
width = 0.35

bars1 = ax1.bar(x - width/2, test_r2_scores, width, label='R² Score', color='skyblue', edgecolor='black')
ax1_twin = ax1.twinx()
bars2 = ax1_twin.bar(x + width/2, test_rmse_scores, width, label='RMSE', color='coral', edgecolor='black')

ax1.set_xlabel('Model', fontsize=12)
ax1.set_ylabel('R² Score', fontsize=12, color='skyblue')
ax1_twin.set_ylabel('RMSE', fontsize=12, color='coral')
ax1.set_title('Model Performance Comparison (Test Set)', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(models)
ax1.tick_params(axis='y', labelcolor='skyblue')
ax1_twin.tick_params(axis='y', labelcolor='coral')
ax1.legend(loc='upper left')
ax1_twin.legend(loc='upper right')
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
            f'{height:.4f}', ha='center', va='bottom', fontsize=9)

# Plot 2: Predictions vs Actual (top-right)
ax2 = fig.add_subplot(gs[0, 1])
# Use XGBoost predictions (best model typically)
ax2.scatter(y_test, xgb_test_pred, alpha=0.5, s=20, color='steelblue', edgecolors='black', linewidth=0.5)

# Add perfect prediction line
min_val = min(y_test.min(), xgb_test_pred.min())
max_val = max(y_test.max(), xgb_test_pred.max())
ax2.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, label='Perfect Prediction')

ax2.set_xlabel(f'Actual {y_test.name}', fontsize=12)
ax2.set_ylabel(f'Predicted {y_test.name}', fontsize=12)
ax2.set_title(f'XGBoost: Predictions vs Actual (R² = {xgb_test_r2:.4f})', fontsize=14, fontweight='bold')
ax2.legend()
ax2.grid(alpha=0.3)

# Plot 3: Feature Importance (bottom-left)
ax3 = fig.add_subplot(gs[1, 0])
top_features = feature_importance.head(10)
ax3.barh(range(len(top_features)), top_features['importance'], color='lightgreen', edgecolor='black')
ax3.set_yticks(range(len(top_features)))
ax3.set_yticklabels(top_features['feature'])
ax3.set_xlabel('Importance', fontsize=12)
ax3.set_ylabel('Feature', fontsize=12)
ax3.set_title('Top 10 Feature Importance (XGBoost)', fontsize=14, fontweight='bold')
ax3.invert_yaxis()
ax3.grid(axis='x', alpha=0.3)

# Plot 4: Residuals Plot (bottom-right)
ax4 = fig.add_subplot(gs[1, 1])
residuals = y_test - xgb_test_pred
ax4.scatter(xgb_test_pred, residuals, alpha=0.5, s=20, color='purple', edgecolors='black', linewidth=0.5)
ax4.axhline(y=0, color='r', linestyle='--', linewidth=2)
ax4.set_xlabel(f'Predicted {y_test.name}', fontsize=12)
ax4.set_ylabel('Residuals (Actual - Predicted)', fontsize=12)
ax4.set_title(f'XGBoost: Residuals Plot (MAE = {xgb_test_mae:.4f})', fontsize=14, fontweight='bold')
ax4.grid(alpha=0.3)

# Overall title
fig.suptitle('Chicago Beach Weather Sensors - Model Results', fontsize=16, fontweight='bold', y=0.995)

# Save figure
plt.savefig('output/q8_final_visualizations.png', dpi=150, bbox_inches='tight')
print("✓ Saved: output/q8_final_visualizations.png")
plt.close()
```

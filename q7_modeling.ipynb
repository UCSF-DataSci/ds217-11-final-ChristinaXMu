```python
# Import libraries
import pandas as pd
import numpy as np
import os
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb

# Ensure output directory exists
os.makedirs("output", exist_ok=True)

# Load Q6 train/test data
X_train = pd.read_csv('output/q6_X_train.csv')
X_test  = pd.read_csv('output/q6_X_test.csv')
y_train = pd.read_csv('output/q6_y_train.csv').squeeze()  # Convert to Series
y_test  = pd.read_csv('output/q6_y_test.csv').squeeze()

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")

# Train Models
# Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

lr_train_pred = lr_model.predict(X_train)
lr_test_pred  = lr_model.predict(X_test)

# XGBoost Regressor
xgb_model = xgb.XGBRegressor(n_estimators=100, max_depth=6, random_state=42)
xgb_model.fit(X_train, y_train)

xgb_train_pred = xgb_model.predict(X_train)
xgb_test_pred  = xgb_model.predict(X_test)

# Evaluate Model Performance
def compute_metrics(y_true, y_pred):
    r2 = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    return r2, rmse, mae

# Linear Regression metrics
lr_train_r2, lr_train_rmse, lr_train_mae = compute_metrics(y_train, lr_train_pred)
lr_test_r2,  lr_test_rmse,  lr_test_mae  = compute_metrics(y_test, lr_test_pred)

# XGBoost metrics
xgb_train_r2, xgb_train_rmse, xgb_train_mae = compute_metrics(y_train, xgb_train_pred)
xgb_test_r2,  xgb_test_rmse,  xgb_test_mae  = compute_metrics(y_test, xgb_test_pred)

# Save Predictions
predictions_df = pd.DataFrame({
    'actual': y_test.values,
    'predicted_linear': lr_test_pred,
    'predicted_xgboost': xgb_test_pred
})
predictions_df.to_csv('output/q7_predictions.csv', index=False)

# Save Model Metrics
with open('output/q7_model_metrics.txt', 'w') as f:
    f.write("MODEL PERFORMANCE METRICS\n========================\n\n")
    
    f.write("LINEAR REGRESSION:\n")
    f.write(f"  Train R²: {lr_train_r2:.4f}\n")
    f.write(f"  Test R²:  {lr_test_r2:.4f}\n")
    f.write(f"  Train RMSE: {lr_train_rmse:.4f}\n")
    f.write(f"  Test RMSE:  {lr_test_rmse:.4f}\n")
    f.write(f"  Train MAE: {lr_train_mae:.4f}\n")
    f.write(f"  Test MAE:  {lr_test_mae:.4f}\n\n")
    
    f.write("XGBOOST:\n")
    f.write(f"  Train R²: {xgb_train_r2:.4f}\n")
    f.write(f"  Test R²:  {xgb_test_r2:.4f}\n")
    f.write(f"  Train RMSE: {xgb_train_rmse:.4f}\n")
    f.write(f"  Test RMSE:  {xgb_test_rmse:.4f}\n")
    f.write(f"  Train MAE: {xgb_train_mae:.4f}\n")
    f.write(f"  Test MAE:  {xgb_test_mae:.4f}\n")

# Feature Importance (XGBoost)
feature_importance = pd.DataFrame({
    'feature': X_train.columns,
    'importance': xgb_model.feature_importances_
}).sort_values('importance', ascending=False)

feature_importance.to_csv('output/q7_feature_importance.csv', index=False)

# Print Confirmation
print("Q7 artifacts saved:")
print(" - output/q7_predictions.csv")
print(" - output/q7_model_metrics.txt")
print(" - output/q7_feature_importance.csv")
```
